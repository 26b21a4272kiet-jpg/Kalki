```python

import numpy as np

print("=" * 65)
print("           LOAN PAYMENT MATRIX MODEL (F7)")
print("=" * 65)

# ---------------------------------------------------------------
# INPUT: 3 LOANS
# ---------------------------------------------------------------

loans = []

for i in range(3):
    print(f"\nEnter details for Loan {i+1}")

    principal = float(input("Initial Principal (₹): "))

    # Accepts both 1 and 1%
    rate_input = input("Annual Interest Rate (%): ")
    annual_rate = float(rate_input.replace("%", "").strip())

    emi = float(input("Monthly EMI (₹): "))

    loans.append([principal, annual_rate, emi])

# ---------------------------------------------------------------
# CREATE TRANSITION MATRIX
# ---------------------------------------------------------------

A = np.zeros((3, 3))

for i in range(3):
    monthly_rate = loans[i][1] / 100 / 12

    # Principal remaining after one month
    A[i][i] = 1 + monthly_rate

# ---------------------------------------------------------------
# CAYLEY-HAMILTON THEOREM
# ---------------------------------------------------------------

I = np.eye(3)

# Characteristic polynomial of A
coeff = np.poly(A)

c2 = coeff[1]
c1 = coeff[2]
c0 = coeff[3]

print("\n" + "=" * 65)
print("TRANSITION MATRIX A")
print("=" * 65)
print(A)

print("\nCharacteristic Polynomial:")
print(f"λ³ + ({c2:.6f})λ² + ({c1:.6f})λ + ({c0:.6f})")

print("\nCayley-Hamilton Relation:")
print(f"A³ + ({c2:.6f})A² + ({c1:.6f})A + ({c0:.6f})I = 0")

# ---------------------------------------------------------------
# CALCULATE A^60
# ---------------------------------------------------------------

A60 = np.linalg.matrix_power(A, 60)

print("\n" + "=" * 65)
print("A^60  (5 YEARS)")
print("=" * 65)
print(A60)

# ---------------------------------------------------------------
# LOAN PAYMENT CALCULATIONS
# ---------------------------------------------------------------

print("\n" + "=" * 65)
print("LOAN RESULTS")
print("=" * 65)

total_interest_all = 0

for i in range(3):

    principal = loans[i][0]
    annual_rate = loans[i][1]
    emi = loans[i][2]

    monthly_rate = annual_rate / 100 / 12

    balance = principal
    total_interest = 0
    month = 0

    while balance > 0 and month < 1200:

        interest = balance * monthly_rate
        payment = min(emi, balance + interest)

        total_interest += interest
        balance = balance + interest - payment
        month += 1

        if balance < 0.01:
            balance = 0

    total_interest_all += total_interest

    print(f"\nLoan {i+1}")
    print("-" * 40)
    print(f"Initial Principal : ₹{principal:,.2f}")
    print(f"Annual Rate       : {annual_rate:.2f}%")
    print(f"Monthly EMI       : ₹{emi:,.2f}")
    print(f"Paid Off In       : {month} months")
    print(f"Total Interest    : ₹{total_interest:,.2f}")
    print(f"Balance After 60 Months: ₹{max(balance,0):,.2f}")

print("\n" + "=" * 65)
print(f"TOTAL INTEREST OF ALL LOANS : ₹{total_interest_all:,.2f}")
print("=" * 65)

print("\nModel completed successfully!")
```
